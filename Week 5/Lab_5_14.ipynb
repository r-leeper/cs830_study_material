# DBSCAN

This notebook is for exploring the use of the DBSCAN algorithm for clustering

```python
from sklearn.datasets import load_iris
import numpy as np
import matplotlib.pyplot as plt

iris = load_iris()
```

```python
iris
```

```python
# Selecting last two columns
features = iris['data']
features = features[:,-2:]
```

```python
features
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan = DBSCAN(min_samples=5, eps=0.2)
dbscan.fit(features)
```

```python
np.unique(dbscan.labels_)
```

```python
plt.scatter(features[:,0], features[:,1], c=dbscan.labels_, marker=".")
```

```python
# Labels of noise points
np.sum(dbscan.labels_ == -1)
```

## Lets try with all 4 features

```python
features_full = iris['data']

dbscan_full = DBSCAN(min_samples=5, eps=0.422)
dbscan_full.fit(features_full)

print(np.unique(dbscan_full.labels_))
print(np.sum(dbscan_full.labels_ == -1))
```
