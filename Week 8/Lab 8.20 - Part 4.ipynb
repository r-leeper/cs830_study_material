## Part 4 - Simplier and copied from Parts 1-3

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input
from tensorflow.keras.optimizers import Adam

model = Sequential([
    Input(shape=(28, 28)),
    Flatten(name="flatten"),

    Dense(
        64,
        activation="elu",
        kernel_initializer="he_uniform",
        name="feature_1"
    ),

    Dense(
        128,
        activation="elu",
        kernel_initializer="he_uniform",
        name="feature_2"
    ),

    Dense(
        64,
        activation="elu",
        kernel_initializer="he_uniform",
        name="feature_3"
    ),

    Dense(
        10,
        activation="softmax",
        name="classifier"
    )
])

optimizer = Adam(
    learning_rate=0.000701371397823176,
    clipnorm=2.0
)

model.compile(
    optimizer=optimizer,
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
from tensorflow.keras.datasets import fashion_mnist

(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

history = model.fit(
    x_train,
    y_train,
    validation_data=(x_test, y_test),
    epochs=20,
    batch_size=128
)
```

```python
model.save("fashion_mnist_model.keras")
```

```python
fashion_model = tf.keras.models.load_model(
    "fashion_mnist_model.keras"
)
```

```python
transfer_model = Sequential([
    Input(shape=(28, 28)),
    fashion_model.get_layer("flatten"),
    fashion_model.get_layer("feature_1"),
    fashion_model.get_layer("feature_2"),
    fashion_model.get_layer("feature_3"),
    Dense(10, activation="softmax", name="mnist_classifier")
])
```

### Compile with frozen layers

```python
for layer in transfer_model.layers[:-1]:
    layer.trainable = False
```

```python
transfer_model.compile(
    optimizer=Adam(
        learning_rate=0.000701371397823176,
        clipnorm=2.0
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
transfer_model.fit(
    x_train,
    y_train,
    validation_data=(x_test, y_test),
    epochs=5,
    batch_size=128
)
```

### Unfreeze layers

```python
for layer in transfer_model.layers:
    layer.trainable = True
```

```python
transfer_model.compile(
    optimizer=Adam(
        learning_rate=1e-5,
        clipnorm=2.0
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
transfer_model.fit(
    x_train,
    y_train,
    validation_data=(x_test, y_test),
    epochs=5,
    batch_size=128
)
```
